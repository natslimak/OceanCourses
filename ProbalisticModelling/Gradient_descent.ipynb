```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
```

```python
fy = lambda x1,x2:  x1**2 - 2*x1 + 1.5*x2**2 + 2*x1*x2
x1,x2 = np.meshgrid(np.arange(-20,20,0.1),np.arange(-20,20,0.1))
yout = fy(x1,x2)
fig,ax = plt.subplots(1,1,figsize = (6,6))
ax.contourf(x1,x2,yout)


```

```python
gradf = lambda x1,x2: [2*x1 - 2 + 2*x2, 3*x2 + 2*x1]

# While loop
go = 1  
iter = 0
alpha = 0.1 # learning rate
x0 = [-15,-15] # starting value for X
xin = x0 # in the first iteration, the candidate optimal point is the starting point
fin = fy(xin[0],xin[1])
eps = 1e-5
fig,ax = plt.subplots(1,1,figsize = (6,6))
ax.contourf(x1,x2,yout)
ax.plot(xin[0],xin[1],'or')
while go == 1:
    iter+=1
    fi = fin
    xi = xin
    fin = fy(xi[0],xi[1]) # compute function value
    gradfi = gradf(xi[0],xi[1]) # compute gradient
    xin = [xi[0] - alpha*gradfi[0],xi[1] - alpha*gradfi[1]] # update optimal point candidate
    ax.plot(xin[0],xin[1],'or')
    if np.abs(xin[0] - xi[0]) + np.abs(xin[1] - xi[1]) <= eps: go = 0 # convergence criterion
    if iter > 100: go = 0 # stopping in case max iterations reached without convergence




plt.show()
```
