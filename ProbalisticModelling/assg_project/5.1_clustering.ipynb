## Method 1: K-Means 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler


def cluster_operating_state(df, state, feature_columns=None, state_column="operating_state", random_state=100, max_k=12, show_plots=True,):
    """ Perform K-means clustering on a selected operating state. """


    feature_columns = ["Hs","Tp", "Vw_mean","Vw_std","RPM","Pitch"]

    # --------------------------------------------------
    # Filter chosen operating state
    # --------------------------------------------------
    df_state = (df.loc[df[state_column] == state].copy().reset_index(drop=True))
    df_state = df_state.dropna(subset=feature_columns)

    # --------------------------------------------------
    # Scale features
    # --------------------------------------------------
    scaler = StandardScaler()
    X = scaler.fit_transform(df_state[feature_columns])

    # --------------------------------------------------
    # Find optimal k
    # --------------------------------------------------
    candidate_k = range(2,min(max_k, len(df_state) - 1) + 1)

    silhouette_scores = []

    for k in candidate_k:
        model = KMeans(n_clusters=k,n_init=20,random_state=random_state )
        labels = model.fit_predict(X)
        score = silhouette_score(X, labels)
        silhouette_scores.append(score)

    selection = pd.DataFrame({"k": list(candidate_k),"silhouette": silhouette_scores,})
    selected_k = int(selection.loc[selection["silhouette"].idxmax(), "k" ])

    # --------------------------------------------------
    # Final K-means
    # --------------------------------------------------
    kmeans = KMeans(n_clusters=selected_k, n_init=50, random_state=random_state)
    df_state["cluster"] = kmeans.fit_predict(X)

    # --------------------------------------------------
    # Representative states
    # --------------------------------------------------

    centers = pd.DataFrame(scaler.inverse_transform(kmeans.cluster_centers_), columns=feature_columns,)
    probabilities = (df_state["cluster"].value_counts(normalize=True).sort_index())
    centers["occurrence_probability"] = (probabilities.values)
    centers["cluster"] = np.arange(selected_k)
    centers = (centers.sort_values("Vw_mean").reset_index(drop=True))

    # --------------------------------------------------
    # Plotting
    # --------------------------------------------------

    if show_plots:
        fig, axes = plt.subplots(1, 3, figsize=(16, 4))

        # Silhouette
        axes[0].plot(selection["k"],selection["silhouette"],marker="o")
        axes[0].axvline(selected_k,color="red",linestyle="--")
        axes[0].set(xlabel="Number of clusters (k)",ylabel="Silhouette score",title=f"State {state}: chosen k={selected_k}")

        # Cluster map
        axes[1].scatter(df_state["Vw_mean"],df_state["Hs"],c=df_state["cluster"],cmap="tab10",s=8,alpha=0.35)
        axes[1].scatter(centers["Vw_mean"],centers["Hs"],color="black",marker="x",s=80)

        axes[1].set(xlabel="Mean wind speed (m/s)",ylabel="Hs (m)", title="Representative states")

        # Occurrence
        axes[2].bar(np.arange(len(centers)), centers["occurrence_probability"])
        axes[2].set( xlabel="Representative state",ylabel="Occurrence probability", title="Cluster probability")

        plt.tight_layout()
        plt.show()

    return {
        "data": df_state,
        "centers": centers,
        "selection": selection,
        "selected_k": selected_k,
        "kmeans": kmeans,
        "scaler": scaler,
    }
```

```python
def gmm_component_selection(X, max_components=20, random_state=100):
    """ Select the optimal number of components for a Gaussian Mixture Model using BIC. """


    k_range = range(1, max_components + 1)
    bic = [GaussianMixture(n_components=k, random_state=random_state).fit(X).bic(X) for k in k_range]

    plt.figure(figsize=(8, 4))
    plt.plot(k_range, bic, 'o-')
    plt.xlabel('Number of components k'); plt.ylabel('BIC (lower is better)')
    plt.title('GMM component selection'); plt.grid(alpha=0.3); plt.show()
```

```python
def cluster_operating_state_gmm(df, state, k_num, feature_columns, random_state=100, show_plots=True):

    # Keep only the requested operating state and usable feature rows
    df_state = df.loc[df["operating_state"].isin(np.atleast_1d(state))].copy()
    feature_data = (df_state[feature_columns].replace([np.inf, -np.inf], np.nan).dropna())
    df_state = df_state.loc[feature_data.index].copy()
    
    # Normalize the features
    scaler = StandardScaler()
    X = scaler.fit_transform(feature_data)

    # Run Gaussian Mixture Model clustering
    gmm = GaussianMixture(n_components=k_num, random_state=random_state)
    df_state["cluster"] = gmm.fit_predict(X)

    # Get cluster centers and occurrence probabilities
    centers = pd.DataFrame(scaler.inverse_transform(gmm.means_),columns=feature_columns,)
    centers.insert(0, "cluster", np.arange(k_num))
    probabilities = (df_state["cluster"].value_counts(normalize=True).reindex(range(k_num), fill_value=0))
    centers["occurrence_probability"] = probabilities.to_numpy()

    # Plot
    if show_plots:
        fig, axes = plt.subplots(1, 2, figsize=(12, 4))

        if "Vw_mean" in feature_columns and "Hs" in feature_columns:
            axes[0].scatter( df_state["Vw_mean"], df_state["Hs"], c=df_state["cluster"], s=8, alpha=0.35, cmap="tab10", ) 
            axes[0].scatter( centers["Vw_mean"], centers["Hs"], c="black", marker="x", s=70, label="GMM mean", ) 
            axes[0].set( xlabel="Mean wind speed (m/s)", ylabel="Significant wave height (m)", title=f"State {state}: {k_num} clusters")
            axes[0].legend()

        centers_sorted = centers.sort_values("Vw_mean")
        axes[1].bar(np.arange(len(centers_sorted)),centers_sorted["occurrence_probability"],)
        axes[1].set(xlabel="Representative state (sorted by wind speed)",ylabel="Occurrence probability",title="Cluster frequencies")

        plt.tight_layout()
        plt.show()

    return {
        "df_state": df_state,
        "centers": centers,
        "model": gmm,
        "scaler": scaler,
        "k_num": k_num,
    }
```

```python
reference = df["operating_state"].eq(0)  # Assumes state 0 means normal operation
results = pd.DataFrame(index=df.index)
fit_time = {}

profile_cols = ["Hs", "Tp", "Vw_mean", "Vw_std", "RPM", "Pitch", "DEL"]

def _align_cluster_labels(labels, data):
    labels = np.asarray(labels)
    if labels.ndim != 1 or len(labels) != len(data):
        raise ValueError("labels must contain one label for each row in data")
    return pd.Series(labels, index=data.index, name="cluster")

def cluster_profile(labels, data=df):
    """Return cluster sizes and median values for this dataset."""
    labels = _align_cluster_labels(labels, data)
    prof = data[profile_cols].groupby(labels).median().round(2)
    sizes = labels.value_counts().reindex(prof.index).astype(int)
    prof.insert(0, "size", sizes)
    return prof

def plot_clusters(labels, title, max_clusters=16, data=df):
    """Plot selected dataset measurements by cluster; DBSCAN noise is shown in black."""
    labels = _align_cluster_labels(labels, data).to_numpy()
    sizes = pd.Series(labels[labels >= 0]).value_counts()
    main = sizes.index[:max_clusters]
    x = data["Vw_mean"].to_numpy()
    plot_specs = [
        ("Hs", "Significant wave height [m]"),
        ("RPM", "RPM"),
        ("DEL", "DEL"),
    ]
    fig, axes = plt.subplots(1, len(plot_specs), figsize=(18, 5))
    for ax, (column, ylabel) in zip(axes, plot_specs):
        y = data[column].to_numpy()
        small = (labels >= 0) & ~np.isin(labels, main)
        if small.any():
            ax.plot(x[small], y[small], ".", ms=2, color="lightgrey", label="other clusters")
        for i, cluster in enumerate(main):
            mask = labels == cluster
            ax.plot(x[mask], y[mask], ".", ms=2, color=f"C{i}", label=f"cluster {cluster} (n={mask.sum()})")
        noise = labels == -1
        if noise.any():
            ax.plot(x[noise], y[noise], "x", ms=3, color="black", label=f"noise (n={noise.sum()})")
        ax.set_xlabel("Mean wind speed [m/s]")
        ax.set_ylabel(ylabel)
        ax.grid(alpha=0.3)
    axes[0].legend(markerscale=5, fontsize=8, loc="best")
    fig.suptitle(title, fontsize=14)
    plt.tight_layout()
    plt.show()

def evaluate_filter(keep, name, data=df):
    """Compare a predicted normal-operation mask with operating_state == 0."""
    keep = np.asarray(keep, dtype=bool)
    if keep.ndim != 1 or len(keep) != len(data):
        raise ValueError("keep must contain one boolean for each row in data")
    actual_normal = data["operating_state"].eq(0).to_numpy()
    true_positive = int((keep & actual_normal).sum())
    precision = true_positive / keep.sum() if keep.sum() else np.nan
    recall = true_positive / actual_normal.sum() if actual_normal.sum() else np.nan
    f1 = 2 * precision * recall / (precision + recall) if precision + recall else np.nan
    return {
        "method": name,
        "kept [%]": round(100 * keep.mean(), 1),
        "precision": round(precision, 3) if np.isfinite(precision) else np.nan,
        "recall": round(recall, 3) if np.isfinite(recall) else np.nan,
        "F1": round(f1, 3) if np.isfinite(f1) else np.nan,
    }

print(f'{reference.mean():.1%} of df rows have operating_state = 0 (the assumed normal state).')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

feature_columns = ["Hs", "Tp", "Vw_mean"]

# Keep complete rows so the feature data, DEL, and eventual labels stay aligned
data = (
    df[feature_columns + ["DEL"]]
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
)

X_scaled = StandardScaler().fit_transform(data[feature_columns])
del_values = data["DEL"].to_numpy()

# A starting value: 2 × number of features
base_min_samples = 2 * len(feature_columns)

neighbors = NearestNeighbors(n_neighbors=base_min_samples).fit(X_scaled)
distances = neighbors.kneighbors(X_scaled)[0][:, -1]
k_distances = np.sort(distances)

plt.figure(figsize=(8, 4))
plt.plot(k_distances)
plt.xlabel("Rows, sorted by distance")
plt.ylabel(f"Distance to {base_min_samples}th nearest neighbor")
plt.title("DBSCAN k-distance plot (look for the elbow)")
plt.grid(True)
plt.show()
```

```python
min_samples_values = [6, 10, 15]
quantiles = np.linspace(0.70, 0.98, 8)
score_rows = []
labels_by_params = {}

for min_samples in min_samples_values:
    neighbors = NearestNeighbors(n_neighbors=min_samples).fit(X_scaled)
    distances = neighbors.kneighbors(X_scaled)[0][:, -1]
    eps_values = np.unique(np.quantile(distances, quantiles))
    eps_values = eps_values[eps_values > 0]

    for eps in eps_values:
        labels = DBSCAN(eps=float(eps), min_samples=min_samples).fit_predict(X_scaled)
        keep = labels != -1
        n_clusters = len(np.unique(labels[keep]))
        labels_by_params[(min_samples, float(eps))] = labels

        if keep.any() and n_clusters > 0:
            scored = pd.DataFrame({
                "cluster": labels[keep],
                "DEL": del_values[keep],
            })
            del_lumped = scored.groupby("cluster")["DEL"].transform("mean")

            rmse = np.sqrt(mean_squared_error(scored["DEL"], del_lumped))
            r2 = r2_score(scored["DEL"], del_lumped)
        else:
            rmse = np.nan
            r2 = np.nan

        score_rows.append({
            "min_samples": min_samples,
            "eps": float(eps),
            "n_clusters": n_clusters,
            "noise_fraction": np.mean(labels == -1),
            "DEL_RMSE": rmse,
            "DEL_R2": r2,
        })

metrics = pd.DataFrame(score_rows)
display(metrics.sort_values("DEL_RMSE").head(15))
```

```python
# One representative sea state (Hs, Tp) per 1 m/s wind-speed bin.
# A one-component GMM gives the bin centroid; DEL is lumped to the bin mean.

wind_bin_width = 1.0
representative_features = ["Hs", "Tp"]

lumping_data = (
    df[["Vw_mean", "Hs", "Tp", "DEL"]]
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
    .copy()
)
lumping_data["wind_bin"] = (
    np.floor(lumping_data["Vw_mean"] / wind_bin_width) * wind_bin_width
)

representative_rows = []
for wind_bin, bin_data in lumping_data.groupby("wind_bin", sort=True):
    gmm = GaussianMixture(n_components=1, random_state=100, reg_covar=1e-6)
    gmm.fit(bin_data[representative_features])
    hs_rep, tp_rep = gmm.means_[0]

    representative_rows.append({
        "wind_bin_low": wind_bin,
        "wind_bin_high": wind_bin + wind_bin_width,
        "Vw_mean": bin_data["Vw_mean"].mean(),
        "Hs_rep": hs_rep,
        "Tp_rep": tp_rep,
        "DEL_lumped": bin_data["DEL"].mean(),
        "n_records": len(bin_data),
    })

lumped_states = pd.DataFrame(representative_rows)
lumping_data = lumping_data.merge(
    lumped_states[["wind_bin_low", "DEL_lumped"]],
    left_on="wind_bin",
    right_on="wind_bin_low",
    how="left",
)

rmse = np.sqrt(mean_squared_error(lumping_data["DEL"], lumping_data["DEL_lumped"]))
r2 = r2_score(lumping_data["DEL"], lumping_data["DEL_lumped"])

print(f"Wind-speed bin width: {wind_bin_width:g} m/s")
print(f"Rows reduced from {len(lumping_data):,} to {len(lumped_states):,} representative sea states")
print(f"DEL lumping RMSE: {rmse:,.2f}")
print(f"DEL lumping R2: {r2:.4f}")
display(lumped_states)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].scatter(lumping_data["Hs"], lumping_data["Tp"], s=8, alpha=0.2, label="Original records")
axes[0].scatter(lumped_states["Hs_rep"], lumped_states["Tp_rep"], color="red", marker="x", s=70, label="Representative state")
axes[0].set(xlabel="Hs (m)", ylabel="Tp (s)", title="One representative (Hs, Tp) per wind bin")
axes[0].legend()

axes[1].plot(lumping_data["Vw_mean"], lumping_data["DEL"], ".", alpha=0.15, label="Original DEL")
axes[1].plot(lumped_states["Vw_mean"], lumped_states["DEL_lumped"], "o-", color="red", label="Bin-mean DEL")
axes[1].set(xlabel="Mean wind speed (m/s)", ylabel="DEL", title="DEL by wind-speed bin")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Compare 1-3 GMM components within each wind-speed bin.
# Require at least 10 records per component to avoid unstable splits.
max_components = 3
min_records_per_component = 10

cluster_evaluation_rows = []
for wind_bin, bin_data in lumping_data.groupby("wind_bin", sort=True):
    X_bin = StandardScaler().fit_transform(bin_data[representative_features])
    supported_max = min(max_components, max(1, len(bin_data) // min_records_per_component))
    bic_by_k = {}
    silhouette_by_k = {}

    for n_components in range(1, supported_max + 1):
        model = GaussianMixture(
            n_components=n_components,
            n_init=5,
            random_state=100,
            reg_covar=1e-5,
        )
        labels = model.fit_predict(X_bin)
        bic_by_k[n_components] = model.bic(X_bin)

        if 1 < len(np.unique(labels)) < len(labels):
            silhouette_by_k[n_components] = silhouette_score(X_bin, labels)
        else:
            silhouette_by_k[n_components] = np.nan

    selected_components = min(bic_by_k, key=bic_by_k.get)
    cluster_evaluation_rows.append({
        "wind_bin": f"{wind_bin:g}-{wind_bin + wind_bin_width:g}",
        "n_records": len(bin_data),
        "candidate_components": supported_max,
        "selected_components_BIC": selected_components,
        "BIC_1": bic_by_k.get(1, np.nan),
        "BIC_2": bic_by_k.get(2, np.nan),
        "BIC_3": bic_by_k.get(3, np.nan),
        "silhouette_selected": silhouette_by_k.get(selected_components, np.nan),
    })

cluster_evaluation = pd.DataFrame(cluster_evaluation_rows)
display(cluster_evaluation)

fig, ax = plt.subplots(figsize=(12, 4))
ax.bar(cluster_evaluation["wind_bin"], cluster_evaluation["selected_components_BIC"])
ax.set(xlabel="Wind-speed bin (m/s)", ylabel="Selected GMM components", ylim=(0, max_components + 0.5), title="Suggested number of (Hs, Tp) representatives per wind bin")
ax.set_yticks(range(1, max_components + 1))
ax.tick_params(axis="x", rotation=70)
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("Selection rule: lowest BIC; candidate splits require at least 10 records per component.")
print("Bins with fewer than 20 records are evaluated as one component only.")
```

```python
def plot_selected_wind_bin(wind_bin_low):
    """Plot (Hs, Tp) observations and GMM representatives for one wind bin."""
    bin_data = lumping_data.loc[lumping_data["wind_bin"] == wind_bin_low].copy()
    if bin_data.empty:
        available_bins = np.sort(lumping_data["wind_bin"].unique())
        raise ValueError(f"No records for bin {wind_bin_low}. Available lower bounds: {available_bins}")

    bin_label = f"{wind_bin_low:g}-{wind_bin_low + wind_bin_width:g}"
    evaluation_row = cluster_evaluation.loc[cluster_evaluation["wind_bin"] == bin_label]
    if evaluation_row.empty:
        raise ValueError(f"No component evaluation found for wind bin {bin_label}.")

    n_components = int(evaluation_row["selected_components_BIC"].iloc[0])
    feature_scaler = StandardScaler()
    X_bin = feature_scaler.fit_transform(bin_data[representative_features])
    model = GaussianMixture(
        n_components=n_components,
        n_init=5,
        random_state=100,
        reg_covar=1e-5,
    )
    labels = model.fit_predict(X_bin)

    representatives = pd.DataFrame(
        feature_scaler.inverse_transform(model.means_),
        columns=representative_features,
    )
    representatives["component"] = np.arange(n_components)
    representatives["records"] = pd.Series(labels).value_counts().reindex(
        range(n_components), fill_value=0
    ).to_numpy()
    representatives["share"] = representatives["records"] / len(bin_data)

    fig, ax = plt.subplots(figsize=(7, 5))
    scatter = ax.scatter(
        bin_data["Hs"],
        bin_data["Tp"],
        c=labels,
        cmap="tab10",
        s=18,
        alpha=0.55,
    )
    ax.scatter(
        representatives["Hs"],
        representatives["Tp"],
        marker="X",
        s=180,
        color="black",
        edgecolor="white",
        linewidth=1.2,
        label="GMM representative",
        zorder=3,
    )
    for row in representatives.itertuples():
        ax.annotate(f"C{row.component}", (row.Hs, row.Tp), xytext=(6, 6), textcoords="offset points")

    ax.set(
        xlabel="Significant wave height, Hs (m)",
        ylabel="Peak period, Tp (s)",
        title=f"Wind bin {bin_label} m/s: {n_components} GMM representative(s), n={len(bin_data)}",
    )
    ax.grid(alpha=0.25)
    ax.legend(loc="best")
    plt.tight_layout()
    plt.show()

    display(representatives)
    return {"model": model, "representatives": representatives, "labels": labels}


# Change this lower-bound value to inspect another wind-speed bin.
selected_wind_bin = 11.0
selected_bin_result = plot_selected_wind_bin(selected_wind_bin)
```

```python
# Compare one representative per wind bin with the BIC-selected GMM representatives.
# DEL predictions are the mean observed DEL within each assigned representative group.
evaluation_data = lumping_data.reset_index(drop=True).copy()
evaluation_data["DEL_pred_1_per_bin"] = evaluation_data["DEL_lumped"]
evaluation_data["DEL_pred_BIC"] = np.nan

gmm_cluster_counts = cluster_evaluation.set_index("wind_bin")["selected_components_BIC"]
for wind_bin, bin_data in evaluation_data.groupby("wind_bin", sort=True):
    bin_label = f"{wind_bin:g}-{wind_bin + wind_bin_width:g}"
    n_components = int(gmm_cluster_counts.loc[bin_label])
    feature_scaler = StandardScaler()
    X_bin = feature_scaler.fit_transform(bin_data[representative_features])
    model = GaussianMixture(
        n_components=n_components,
        n_init=5,
        random_state=100,
        reg_covar=1e-5,
    )
    labels = model.fit_predict(X_bin)
    cluster_del_means = bin_data.assign(_cluster=labels).groupby("_cluster")["DEL"].mean()
    evaluation_data.loc[bin_data.index, "DEL_pred_BIC"] = cluster_del_means.loc[labels].to_numpy()


def tail_error_metrics(actual, predicted):
    errors = predicted - actual
    return {
        "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
        "MAE": np.mean(np.abs(errors)),
        "Mean bias (pred - actual)": np.mean(errors),
        "Underestimated [%]": 100 * np.mean(errors < 0),
    }


sensitivity_rows = []
for cohort, tail_fraction in [("All records", None), ("Top 20% DEL", 0.20), ("Top 10% DEL", 0.10), ("Top 5% DEL", 0.05)]:
    if tail_fraction is None:
        subset = evaluation_data
        cutoff = np.nan
    else:
        cutoff = evaluation_data["DEL"].quantile(1 - tail_fraction)
        subset = evaluation_data.loc[evaluation_data["DEL"] >= cutoff]

    for approach, prediction_column in [
        ("1 representative / bin", "DEL_pred_1_per_bin"),
        ("BIC-selected representatives", "DEL_pred_BIC"),
    ]:
        metrics = tail_error_metrics(subset["DEL"].to_numpy(), subset[prediction_column].to_numpy())
        sensitivity_rows.append({
            "DEL group": cohort,
            "Actual DEL cutoff": cutoff,
            "Rows": len(subset),
            "Approach": approach,
            **metrics,
        })

sensitivity_results = pd.DataFrame(sensitivity_rows)
display(
    sensitivity_results.style.format({
        "Actual DEL cutoff": "{:.0f}",
        "RMSE": "{:.0f}",
        "MAE": "{:.0f}",
        "Mean bias (pred - actual)": "{:+.0f}",
        "Underestimated [%]": "{:.1f}%",
    }).set_table_styles([
        {"selector": "th, td", "props": [("border", "1px solid #e6e6e6")]},
        {"selector": "th", "props": [("background-color", "#f5f5f5")]},
    ])
)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
cohort_order = ["All records", "Top 20% DEL", "Top 10% DEL", "Top 5% DEL"]
for approach in ["1 representative / bin", "BIC-selected representatives"]:
    approach_results = sensitivity_results.loc[sensitivity_results["Approach"] == approach].set_index("DEL group").loc[cohort_order]
    axes[0].plot(cohort_order, approach_results["RMSE"], marker="o", label=approach)
    axes[1].plot(cohort_order, approach_results["Underestimated [%]"], marker="o", label=approach)

axes[0].set(ylabel="DEL RMSE", title="Error in the high-damage tail")
axes[1].set(ylabel="Actual DEL underestimated (%)", title="Underestimation frequency")
for ax in axes:
    ax.set_xlabel("Evaluation subset")
    ax.grid(alpha=0.3)
    ax.tick_params(axis="x", rotation=20)
    ax.legend()
plt.tight_layout()
plt.show()

print(f"BIC-selected representatives total: {int(cluster_evaluation['selected_components_BIC'].sum())}")
print("Note: DEL group means are evaluated on the same records used to form each group (in-sample lumping error).")
```

```python
# SHAP sensitivity of DEL to Hs, Tp, and mean wind speed.
# This explains a predictive surrogate; it is not a causal or fatigue-damage model.
import shap
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

sensitivity_features = ["Hs", "Tp", "Vw_mean"]
sensitivity_data = (
    df[sensitivity_features + ["DEL"]]
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
    .copy()
)

X_sensitivity = sensitivity_data[sensitivity_features]
y_sensitivity = sensitivity_data["DEL"]
X_train, X_test, y_train, y_test = train_test_split(
    X_sensitivity,
    y_sensitivity,
    test_size=0.2,
    random_state=100,
)

sensitivity_model = RandomForestRegressor(
    n_estimators=200,
    min_samples_leaf=5,
    random_state=100,
    n_jobs=-1,
)
sensitivity_model.fit(X_train, y_train)
y_test_pred = sensitivity_model.predict(X_test)

holdout_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
holdout_r2 = r2_score(y_test, y_test_pred)
print(f"Holdout R2: {holdout_r2:.4f}")
print(f"Holdout RMSE: {holdout_rmse:,.0f}")

# Explain a reproducible subset of the held-out rows to keep computation practical.
explain_size = min(2000, len(X_test))
X_explain = X_test.sample(n=explain_size, random_state=100)
explainer = shap.TreeExplainer(sensitivity_model)
shap_values = explainer.shap_values(X_explain)

mean_abs_shap = np.abs(shap_values).mean(axis=0)
shap_summary = (
    pd.DataFrame({
        "Feature": sensitivity_features,
        "Mean absolute SHAP (DEL units)": mean_abs_shap,
    })
    .sort_values("Mean absolute SHAP (DEL units)", ascending=False)
    .reset_index(drop=True)
)

display(
    shap_summary.style.format({"Mean absolute SHAP (DEL units)": "{:,.0f}"})
    .set_table_styles([
        {"selector": "th, td", "props": [("border", "1px solid #e6e6e6")]},
        {"selector": "th", "props": [("background-color", "#f5f5f5")]},
    ])
)

fig, ax = plt.subplots(figsize=(7, 4))
shap.summary_plot(shap_values, X_explain, plot_type="bar", show=False)
ax = plt.gca()
ax.set_xlabel("Mean absolute SHAP value (DEL units)")
ax.set_title("Global DEL sensitivity to sea-state and wind inputs")
plt.tight_layout()
plt.show()

print("Interpretation: larger mean absolute SHAP values indicate greater contribution to the surrogate's DEL predictions.")
print("Inputs Hs, Tp, and Vw_mean are correlated, so individual SHAP contributions are predictive attributions, not independent causal effects.")
```
