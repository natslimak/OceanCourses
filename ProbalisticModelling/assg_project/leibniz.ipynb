## Data Investigation of the Leibniz University Hannover Dataset

The investigated wind turbine is the NREL 5MW reference turbine (Jonkman et al., 2009) on the OC3 monopile (Jonkman and Musial, 2010).  
Environmental conditions are based on statistical distributions by HÃ¼bler et al. (2018), which were fitted to the measured environmental conditions at the FINO 3 research platform.  
soil-structure-interaction (HÃ¤fele et al., 2016; HÃ¼bler et al., 2018). 
The time series of the bending moments at mudline and blade root are computed using an adapted version of the FASTv8 simulation code by the NREL (Jonkman, 2013) considering  
10,000 aeroelastic simulations for varying environmental conditions (and varying random wind and wave seeds) were conducted.  
Short-term damage equivalent loads (DELs) representing fatigue were calculated for several relevant positions (at mudline and at the blade root) 
For further information, it is referred to "Schmidt, F., HÃ¼bler, C., and Rolfes, R.: Kriging meta-models for damage equivalent load assessment of idling offshore wind turbines,  
Wind Energ. Sci. Discuss. [preprint], https://doi.org/10.5194/wes-2025-83, in review, 2025. 

Link to the website:
https://data.uni-hannover.de/dataset/simulation-data-and-meta-models-of-an-offshore-wind-turbine

```python
import pandas as pd

path = r"data\sim_data.txt"

df = pd.read_csv(
    path,
    sep=r"\s+",
    skiprows=2,
    header=None,
)

df.columns = [
    "vs [m/s]", "TI [percent]", "Hs [m]", "Tp [s]", "WWM [degree]",
    "Fx_DEL [kN]", "My_DEL [kNm]", "Fy_DEL [kN]", "Mx_DEL [kNm]",
    "MRoot,edge_DEL [kN]", "MRoot,edge_DEL [kN*m]",
    "MRoot,flap_DEL [kN]", "MRoot,flap_DEL [kN*m]",
]

df
```

```python
# Determine amount of rows in the dataframe
num_rows = len(df)
print(f"Number of rows in the dataframe: {num_rows}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(df["vs [m/s]"], bins=40, color='blue', alpha=0.7)
plt.title("Histogram of vs [m/s]")
plt.xlabel("vs [m/s]")
plt.ylabel("Frequency")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

axes[0].hist(df["Fx_DEL [kN]"], bins=40, color="blue", alpha=0.7)
axes[1].hist(df["Fy_DEL [kN]"], bins=40, color="red", alpha=0.7)
axes[2].hist(df["Mx_DEL [kNm]"], bins=40, color="green", alpha=0.7)
axes[3].hist(df["My_DEL [kNm]"], bins=40, color="orange", alpha=0.7)

axes[0].set_title("Distribution of Fx_DEL")
axes[1].set_title("Distribution of Fy_DEL")
axes[2].set_title("Distribution of Mx_DEL")
axes[3].set_title("Distribution of My_DEL")

plt.tight_layout()
plt.show()
```
