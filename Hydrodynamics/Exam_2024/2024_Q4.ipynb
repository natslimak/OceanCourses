# Problem 4: Flow around a Cylinder


```python
import numpy as np
import sympy as sp
from scipy.optimize import fsolve
import matplotlib.pyplot as plt
from matplotlib import animation
```

```python
d = 0.2                     # Cylinder Diameter [m]
U_0 = 0.65                  # Flow velocity [m/s]
viscosity = 1.3*10**-6      # Kinematic viscosity [m2/s]
rho_water = 1000            # density of water (kg/m^3)
```

### Question 4.1



```python
Re = (U_0 * d) / viscosity    # Reynolds number
C_D = 1.2                     # Drag coefficient for a cylinder in laminar flow (Re > 10000)

F = 0.5 * rho_water * U_0**2 * C_D * d  # Force per unit length P11 p.15

print(f"Reynolds number: {Re:.0f}")
print(f"Drag force on the cylinder: {F:.0f} N")
```

### Question 4.2


```python
S = 0.2                   # Strouhal number
f = S * U_0 / d           # Vortex shedding frequency [Hz]

print(f"Vortex shedding frequency: {f:.2f} Hz")
```
