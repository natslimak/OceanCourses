#### Question 5.1


```python
U_f = np.sqrt(shear_0/rho_water)  # Friction velocity [m/s]

print(f"Friction velocity U_f: {U_f:.3f} m/s")
```

#### Question 5.2


```python
x1 = 100
y1 = 1 # height [m]

delta  = lambda x: (kappa ** 2 * x) / 2
delta_1 = delta(x1)

U_aty = lambda delta, y: 1/kappa *np.log(delta/y) * U_f

print(f"Velocity at y={y1} m = {U_aty(delta_1, y1):.2f} m/s")
```
