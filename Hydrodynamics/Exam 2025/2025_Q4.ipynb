```python
# Given data
g = 9.81                    # gravitational acceleration (m/s^2)
viscosity = 1.3*10**-6      # kinematic viscosity of water (m^2/s)    
rho_water = 1000            # density of water (kg/m^3)
kappa = 0.4                 # von Karman constant
rho_air = 1.2               # air density (kg/m^3)
viscosity_air = 1.5*10**-5  # kinematic viscosity of air (m^2/s) 

d = 0.1 * 10**-3
rho_s = 2650
```

#### Question 4.1


```python
F_g = g * rho_s * np.pi/6 * d**3
print(f"Gravitational force: {F_g:.1e}")
```

#### Question 4.2


```python
F_b = g * rho_water * np.pi/6 * d**3
print(f"Buoyancy force: {F_b:.1e}")
```

#### Question 4.3


```python
F_d = g * (np.pi * d**3 / 6) * (rho_s - rho_water)
print(f"Drag force: {F_d:.1e}")
```

#### Question 4.4


```python
s = rho_s / rho_water
w_s = g * d**2 * (s-1) / (18 * viscosity)
print(f"Settling velocity: {w_s:.4f} m/s")
```
