```python
import numpy as np
import sympy as sp
from scipy.optimize import fsolve
import matplotlib.pyplot as plt
from matplotlib import animation
```

```python
a = 0.08                    # amplitude [m]
viscosity = 1.3*10**-6      # Kinematic viscosity [m2/s]
rho_water = 1000            # density of water (kg/m^3)
g = 9.81                    # gravitational acceleration [m/s2]
kappa = 0.4                 # von Karman constant
k = 1.2                     # wave number [1/m]
D = 1.5                     # water depth [m]
u_0 = 0.65                  # free stream velocity [m/s]

# Shallow-water, linear, irrotational
```

#### Question 1


```python
def determine_flow_type(velocity, depth):
    Froude_number = velocity / np.sqrt(g * depth)

    if Froude_number < 1:
        flow_type = "Subcritical"
    else:
        flow_type = "Supercritical"

    return flow_type, Froude_number

flow_type, Fr = determine_flow_type(u_0, D)

# Bed perturbation z_b = a * cos(kx) <--- given in task
# amplitude of perturbation velocity h = -z_b / (1-Fr^2)

# Compute perturbation velocity amplitude
u_amp = (u_0 / D) * (a / (1 - Fr**2))
print(f"Perturbation velocity amplitude: {u_amp:.4f} m/s")
```
