# Unsteady Wind-Driven Turbulent Boundary Layer

```python
import numpy as np
import sympy as sp
from scipy.optimize import fsolve
import matplotlib.pyplot as plt
from matplotlib import animation
```

```python
shear_0 = 0.15              # Shear at water surface [m/s]
viscosity = 1.3*10**-6      # Kinematic viscosity [m2/s]
rho_water = 1000            # density of water (kg/m^3)
kappa = 0.4                 # von Karman constant
```

#### Question 1
What is the friction velocity?

```python

U_f = np.sqrt(shear_0/rho_water)  # Friction velocity [m/s]

print(f"Friction velocity U_f: {U_f:.4f} m/s")
```

#### Question 2
What is the velocity flow at t=20s and y=0.2m?

```python
t1 = 20 # time [s]
y1 = 0.5 # height [m]

delta_1  = kappa * U_f * t1

U_aty = lambda delta, y: 1/kappa *np.log(delta/y) * U_f

print(f"Velocity at y={y1} m and t={t1} s: U(y,t) = {U_aty(delta_1, y1):.4f} m/s")
```
