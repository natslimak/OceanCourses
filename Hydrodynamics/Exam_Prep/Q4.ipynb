# Longshore Curent in Surf Zone


```python
import numpy as np
import sympy as sp
from scipy.optimize import fsolve
import matplotlib.pyplot as plt
from matplotlib import animation
```

```python
H_b = 0.9                       # wave hegiht [m]
D_b = 1.2                       # water depth [m]
T = 7.0                         # wave period [s]
alpha_0 = 25.0 * np.pi/180.0    # wave angle [rad]
H_0 = 1.0                       # deep-water wave height [m]
viscosity = 1.3*10**-6          # Kinematic viscosity [m2/s]
rho_water = 1000                # density of water (kg/m^3)
g = 9.81                        # gravitational acceleration [m/s2]
C_f = 0.006                     # bed friction coefficient
```

#### Question 1


```python
# Deep water wave speed
c_0 = g*T/(2.0*np.pi)

# Shallow water celerity at breaking point
c_b = np.sqrt(g * D_b)

# Snell's law to find deep-water wave angle: c_0 / sin(alpha_0) = c_b / sin(alpha_b)
sin_alpha_b = (c_0 / c_b) * np.sin(alpha_0)

if sin_alpha_b > 1.0:
    sin_alpha_b = 1.0  # Limit to valid range
alpha_b = np.arcsin(sin_alpha_b)   # in degrees

print(f"Deep Water celerity c_0 = {c_0:.2f} m/s")
print(f"Shallow-water wave celerity at breaking point c_b = {c_b:.2f} m/s")
print(f"Shallow-water wave angle alpha_b = {alpha_b* 180.0/np.pi:.2f} degrees")
```

#### Question 2


```python
# Radiation strees
S_xy = 0.5 * rho_water * g * H_b**2 * np.cos(alpha_b) * np.sin(alpha_b)
print(f"Radiation stress S_xy = {S_xy:.2f} N/m")
```

#### Question 3


```python
gradient = S_xy / D_b
print(f"Radiation stress gradient dS_xy/dx = {gradient:.2f} N/m²")
```

#### Question 4


```python
# Momentum balance
U_ls = np.sqrt( (S_xy / (rho_water * D_b * C_f)) )
print(f"Longshore current velocity U_ls = {U_ls:.2f} m/s")
```
